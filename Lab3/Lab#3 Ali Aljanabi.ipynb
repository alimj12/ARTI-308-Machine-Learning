# Lab 3: Exploratory Data Analysis (EDA)

EDA is the first and most important step in any Machine Learning project.
Before building models, we must understand:

- What does the data represent?
- Are there missing values?
- Are there outliers?
- What patterns exist?
- Which variables influence others?

If we do not understand the data, we cannot build a good model.

### Why EDA is Important

In real-world machine learning projects:

    1- 70–80% of the time is spent on understanding and cleaning data
    2- Only 20–30% is spent building models

**EDA helps us:**

- Detect errors
- Identify trends
- Discover relationships
- Make business decisions

```python
df['Time'] = pd.to_numeric(df['Time'], errors='coerce')
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
df.dtypes[['Time', 'Amount']]

df.dtypes
```

### Descriptive summary Statistics

```python
# Statistical summary
df.describe(include='all')
```

### Univariate Analysis

```python
filtered_df = df[df['Amount'] <= 1000]
plt.figure(figsize=(10, 6))

sns.histplot(filtered_df['Amount'], bins=20, color='#6b7eb3', edgecolor='white')
plt.title("Distribution of Transaction Amounts")
plt.xlabel("Amount ($)")
plt.ylabel("Count")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

- Shows the distribution of transaction amounts.

- Right skew indicates that most transactions are for small amounts, with a few extreme

## Bivariate Analysis

### Total Amount by Transaction Class

```python
class_revenue = df.groupby('Class')['Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 6))
class_revenue.plot(kind='bar', color=['#6b7eb3', '#d9534f'])

plt.title("Total Amount: Legitimate (0) vs Fraudulent (1)")
plt.xlabel("Class (0: Normal, 1: Fraud)")
plt.ylabel("Total Amount")
plt.xticks(ticks=[0, 1], labels=['Legitimate', 'Fraud'], rotation=0)

plt.show()

print(class_revenue)
```


### Average Transaction Amount by Class

```python
class_avg_amount = df.groupby('Class')['Amount'].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 6))
class_avg_amount.plot(kind='bar', color=['#d9534f', '#6b7eb3'])

plt.title("Average Transaction Amount: Fraud vs Legitimate")
plt.xlabel("Class (0: Normal, 1: Fraud)")
plt.ylabel("Average Amount ($)")
plt.xticks(ticks=[0, 1], labels=['Fraud', 'Legitimate'], rotation=0)

plt.show()

print(class_avg_amount)
```

### Bivariate Analysis: Time and Amount with Class Differentiation

```python
plt.figure(figsize=(12, 6))

sns.scatterplot(data=df[df['Class'] == 0], x='Time', y='Amount',
                color='#6b7eb3', alpha=0.3, label='Legitimate')

sns.scatterplot(data=df[df['Class'] == 1], x='Time', y='Amount',
                color='#d9534f', alpha=1.0, s=50, edgecolor='black', label='Fraud')

plt.title("Transaction Amount vs Time (Highlighting Fraud)")
plt.xlabel("Time (Seconds)")
plt.ylabel("Amount ($)")
plt.legend()
plt.show()
```

### Correlation Matrix

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['Time', 'Amount', 'Class']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix (Time, Amount, Class)")
plt.show()
```

- Correlation close to 1 = strong positive relationship
- Close to 0 = weak relationship

## Time-Based Analysis
### Fraud Distribution Over Time (Seconds)

```python
plt.figure(figsize=(10, 5))

df[df['Class'] == 1]['Time'].hist(bins=50, color='#d9534f', alpha=0.7)

plt.title("Distribution of Fraudulent Transactions Over Time")
plt.xlabel("Time (Seconds from first transaction)")
plt.ylabel("Number of Fraudulent Transactions")
plt.grid(axis='y', alpha=0.3)

plt.show()
```
